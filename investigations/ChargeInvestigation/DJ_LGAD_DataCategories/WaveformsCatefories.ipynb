```python
rows = 2
cols = len(filenames)
size = 5
n_sections = 20
n_events = 20

summ_fig, summ_ax = plt.subplots(1, 1, figsize = (1.5*size, 1.5*size))
output_folder = f"Scan_{n_sections}_{n_events}"
os.system(f"mkdir -p {output_folder}")
amp_means = {}

for f, file in enumerate(filenames):
    amp_means[keys[f]] = []

for r in range(n_sections):
    fig, ax = plt.subplots(rows, cols, figsize = (cols*size, rows*size))
    for f, file in enumerate(filenames):
        print(file)
        arrays = uproot.open(file)["pulse"].arrays()
        channels = arrays["channel"]
        time = arrays["time"]
        baseline_RMS = arrays["baseline_RMS"]
        amp = arrays["amp"]
        timestamps_10 = arrays["LP2_10"]
        timestamps_80 = arrays["LP2_80"]
        # for i, ch in enumerate(channels[0]):
        # y = (i) % cols
        # x = (i) // cols
        counter = 0
        amplitude = amp[:, 5][r*n_events:(r+1)*n_events]
        if (len(amplitude)==0): continue
        amp_means[keys[f]].append(np.mean(amplitude))
        ax[1][f].hist(amplitude, histtype = "step", bins = 30, linewidth = 2, range = (0, 100), label = f"Mean: {np.round(np.mean(amplitude), 3)}")
        ax[1][f].set_xlabel("Amplitude [mV]")
        ax[1][f].set_xlabel("Events")
        ax[1][f].legend()
        for e in range(len(channels[r*n_events:(r+1)*n_events])):
            if baseline_RMS[e][5] > 3: continue
            # if amp[e][3] / (amp[e][2] + amp[e][3] + amp[e][4]) < 0.5: continue
            # if timestamps_10[e][3] < -13 or timestamps_10[e][3] < -13 or timestamps_10[e][3] > -1 or timestamps_10[e][3] > -1: continue
            # ax[f].plot(time[e][0], channels[e][7])
            # if f != 3: ax[f].set_xlim(-12, -10)
            # else: ax[f].set_xlim(-9.5, -7)
            
            ax[0][f].plot(time[e][0], channels[e][5])
            ax[0][f].set_xlim(-7, -2)
            ax[0][f].set_ylim(-160, 30)    
            # if f != 3: ax[0][f].set_xlim(-8, -5.5)
            # else: ax[0][f].set_xlim(-5.0, -1.5)
            ax[0][f].set_title(legend_labels[f])
            ax[0][f].set_xlabel("Time [ns]")
            ax[0][f].set_ylabel("Amplitude [mV]")
            ax[0][f].plot(np.linspace(-7, -2, 2), [-20]*2, "-", color = "black")

            counter+=1
        print(counter)
        fig.savefig(f"{output_folder}/Section_{r}.png")

for f, file in enumerate(filenames):
    summ_ax.plot(amp_means[keys[f]], label = legend_labels[f])
    summ_ax.set_xlabel("Section number")
    summ_ax.set_ylabel("Mean amplitude [mV]")
    summ_ax.legend()
summ_fig.savefig(f"{output_folder}/Summary.png")
# plt.show()
```
