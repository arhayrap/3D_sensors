```python
import uproot
import awkward as ak
import numpy as np
import matplotlib.pyplot as plt
```

```python
base = "/Users/PITLab/Downloads/"
filenames = [
    f"{base}prep_RT_3x3_vb110v_mcp4300.root",    
    f"{base}prep_RT_3x3_vb113v_mcp4300.root",    
]

legend_labels = []

for f in filenames:
    print(f)
    bias_voltage = f.split("_vb")[1].split("v_")[0] + "V"
    wafer = f.split(f"{base}prep_")[1].split("_")[0]
    tag = f.split(f"{base}prep_RT_")[1].split("_")[0]
    legend_labels.append(wafer+" "+bias_voltage+" "+tag)

print(legend_labels)
```

```python
rows = 1
cols = 2 # len(filenames)
size = 5
# for r in range(20):

channels = []
time = []
baseline_RMS = []
amp = []
timestamps_10 = []
timestamps_80 = []
timestamps_90 = []

colors = ["r", "b"]

fig, ax = plt.subplots(rows, cols, figsize = (cols*size, rows*size))
for f, file in enumerate(filenames):
    print(file)
    arrays = uproot.open(file)["pulse"].arrays()
    channels.append(arrays["channel"])
    time.append(arrays["time"])
    baseline_RMS.append(arrays["baseline_RMS"])
    amp.append(arrays["amp"])
    timestamps_10.append(arrays["LP2_10"])
    timestamps_80.append(arrays["LP2_80"])
    timestamps_90.append(arrays["LP2_90"])

print(amp[0])

for f, file in enumerate(filenames):
    cuts = (amp[f][:,1]<600)&(amp[f][:,1]>20)&(amp[f][:,0]<120)
    
    ax[0].hist(amp[f][:,0][cuts], bins = 60, color = colors[f], range = (np.min(amp[:,0]), np.max(amp[:,0])), label = legend_labels[f], histtype = "step")
    ax[0].set_title("Trigger strip")
    ax[0].set_xlabel("Amplitude (tr channel) [mV]")

    ax[1].hist(amp[f][:,1][cuts], bins = 60, color = colors[f], range = (np.min(amp[:,1]), np.max(amp[f][:,1])), label = legend_labels[f], histtype = "step")
    ax[1].set_title("MCP")
    ax[1].set_xlabel("Amplitude (MCP channel) [mV]")

ax[0].legend()
ax[1].legend()
plt.show()
```
